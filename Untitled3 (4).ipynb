```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error, r2_score
```

```python
df = pd.read_csv("/battery_dataset.csv")
print("Number of Rows:", df.shape[0])
print("Number of Columns:", df.shape[1])

df.head()
```

```python
df.info()
```

```python
df.describe()
```

```python
print(df.isna().sum())
```

```python
print("Duplicated rows:", df.duplicated().sum())
```

```python
df = df.drop(columns=["BatteryID"])

df.head()
```

```python
plt.scatter(df["Cycle"], df["SOH"], alpha=0.5, s=5)
plt.xlabel("Cycle")
plt.ylabel("State of Health (SOH)")
plt.title("Battery Cycle vs State of Health")
plt.show()
```

```python
plt.scatter(df["InternalResistance"], df["SOH"], alpha=0.5, s=5)
plt.xlabel("Internal Resistance")
plt.ylabel("State of Health (SOH)")
plt.title("Internal Resistance vs State of Health")
plt.show()
```

```python
plt.scatter(df["Capacity"], df["SOH"], alpha=0.5, s=5)
plt.xlabel("Capacity")
plt.ylabel("State of Health (SOH)")
plt.title("Battery Capacity vs State of Health")
plt.show()
```

```python
correlation = df.corr(numeric_only=True)

correlation["SOH"].sort_values()
```

```python
X1 = df[["Cycle"]]
y = df["SOH"]

X1_train, X1_test, y1_train, y1_test = train_test_split(
    X1, y, test_size=0.2, random_state=46
)

model1 = LinearRegression()
model1.fit(X1_train, y1_train)

predictions1 = model1.predict(X1_test)

mae1 = mean_absolute_error(y1_test, predictions1)
r2_1 = r2_score(y1_test, predictions1)

print("Model 1 - Cycle Only")
print("MAE:", round(mae1, 2))
print("R²:", round(r2_1, 3))
```

```python
X2 = df[["Cycle", "InternalResistance"]]
y = df["SOH"]

X2_train, X2_test, y2_train, y2_test = train_test_split(
    X2, y, test_size=0.2, random_state=46
)

model2 = LinearRegression()
model2.fit(X2_train, y2_train)

predictions2 = model2.predict(X2_test)

mae2 = mean_absolute_error(y2_test, predictions2)
r2_2 = r2_score(y2_test, predictions2)

print("Model 2 - Cycle + Internal Resistance")
print("MAE:", round(mae2, 2))
print("R²:", round(r2_2, 3))
```

```python
X3 = df[["Cycle", "InternalResistance", "Capacity"]]
y = df["SOH"]

X3_train, X3_test, y3_train, y3_test = train_test_split(
    X3, y, test_size=0.2, random_state=46
)

model3 = LinearRegression()
model3.fit(X3_train, y3_train)

predictions3 = model3.predict(X3_test)

mae3 = mean_absolute_error(y3_test, predictions3)
r2_3 = r2_score(y3_test, predictions3)

print("Model 3 - Cycle + Internal Resistance + Capacity")
print("MAE:", round(mae3, 2))
print("R²:", round(r2_3, 3))
```

```python
plt.scatter(y3_test, predictions3, alpha=0.5, s=5)
plt.xlabel("Actual SOH")
plt.ylabel("Predicted SOH")
plt.title("Actual vs Predicted SOH")
plt.show()
```

```python
comparison = pd.DataFrame({
    "Model": [
        "Cycle Only",
        "Cycle + Internal Resistance",
        "Cycle + Internal Resistance + Capacity"
    ],
    "MAE": [
        mae1,
        mae2,
        mae3
    ],
    "R2": [
        r2_1,
        r2_2,
        r2_3
    ]
})

comparison
```
